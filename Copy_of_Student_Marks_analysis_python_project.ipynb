Student Marks Analysis - python project

This project analyses students' marks using python. It calculates total marks, average marks, highest and lowest marks, identifies the top student, and visualisation marksusing a bar chart.

---



```python
1. Importing Libraries and Creating Dataset
```

```python
# student marks analysis project
```

```python
marks = [85, 90, 78, 92, 88]
print(marks)
```

```python
total = sum(marks)
total
```

```python
average = sum(marks)/ len(marks)
average
```

```python
highest = max(marks)
lowest = min(marks)
highest, lowest
```

```python
students = ["Ali", "Sara", "Ahmad", "zara", "usman"]
students
```

```python
3. Finding Highest, LowestMarks and Top student
```

```python
for i in range(len(students)):
  print(students[i], "got", marks[i], "marks")
```

```python
top_index  = marks.index(highest)
top_student = students[top_index]
top_student
```

```python
2. Calculating Total and Average Marks
```

```python
print("student Marks analysis")
print("====================")
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

```python
4. Data Visualisation using bar chart
```

```python
import matplotlib.pyplot as plt
plt.bar(students, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks Analysis")
plt.show()
#
```
